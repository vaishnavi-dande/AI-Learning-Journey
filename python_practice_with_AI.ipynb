```python
# prompt: basic calculator

def calculator():
    """Performs basic arithmetic operations."""

    while True:
        try:
            num1 = float(input("Enter first number: "))
            op = input("Enter operator (+, -, *, /): ")
            num2 = float(input("Enter second number: "))

            if op == "+":
                result = num1 + num2
            elif op == "-":
                result = num1 - num2
            elif op == "*":
                result = num1 * num2
            elif op == "/":
                if num2 == 0:
                    print("Error: Division by zero")
                    continue  # Go back to the beginning of the loop
                result = num1 / num2
            else:
                print("Invalid operator")
                continue  # Go back to the beginning of the loop

            print(f"Result: {result}")
            break  # Exit the loop if the calculation is successful

        except ValueError:
            print("Invalid input. Please enter numbers only.")
        except Exception as e:
            print(f"An error occurred: {e}")


if __name__ == "__main__":
    calculator()
```

```python
#data cleaning and preprocessing
# handling the missing data
# 3 ways to handole the missing values 1.removing them 2.filling them with mean/median/mode 3.forward or backward fill
import pandas as pd
import numpy as np
data={
    "name":["arav","vaishu","nikki","sweety"],
    "age":[20,19,np.nan,27],
    "salary":[np.nan,60000,70000,np.nan]
}

df=pd.DataFrame(data)
print("the data of the salaries of my ppl:",df)
# removing the missing values and filling them with mean/med/mode= rowise
df_cleaned=df.dropna()
print("the dataframe after removing the missing values:",df_cleaned)
df_filled=df.fillna(df.mean(numeric_only=True))
print("the datadrame after filling the missing values:",df_filled)
```

```python
# handlimg the outliers to prevent the irregular predictions of ai-2 methods
# m1-IQR(inter quartile range) method
# m2-z-score method(standard deviation method)
df=pd.DataFrame({"salary":[5000,4000,3000,6500,1000000,15000000,20,100,7000,40]})
print("salary with outliers:",df)
Q1=df["salary"].quantile(0.25)
Q3=df["salary"].quantile(0.75)
IQR=Q2-Q1
lower_bound=Q1-0.0*IQR
upper_bound=Q3+3.0*IQR
df_cleaned=df[((df["salary"]>lower_bound)&(df["salary"]<upper_bound))]
print("the dataframe whithout outliers is:",df_cleaned)
df_outliers=df[((df["salary"]<lower_bound)|(df["salary"]>upper_bound))]
print("the outliers of salary df is:",df_outliers)
print(f"Q1:{Q1}, Q3:{Q3}, IQR:{IQR}")
print(f"lower_bound:{lower_bound}, upper_bound:{upper_bound}")
```
